```python
# Question 1: S&P 500 Additions Analysis (fixed version)
# Google Colab friendly

import requests
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Fetch the Wikipedia page
url = 'https://en.wikipedia.org/wiki/List_of_S%26P_500_companies'
headers = {'User-Agent': 'Mozilla/5.0'}
response = requests.get(url, headers=headers)

# Step 2: Read tables using pandas
tables = pd.read_html(response.text)

# The first table is the list of S&P 500 companies
df = tables[0]

# Step 3: Inspect column names
print("Columns in table:", df.columns.tolist())

# Step 4: Normalize column names
df.columns = [col.strip() for col in df.columns]

# Some Wikipedia versions call it "Date first added" or "Date first added to index"
possible_date_cols = [c for c in df.columns if "Date" in c and "added" in c.lower()]
date_col = possible_date_cols[0] if possible_date_cols else None

# Step 5: Extract relevant columns safely
if date_col:
    df = df[['Symbol', 'Security', date_col]]
    df.rename(columns={date_col: 'Date first added'}, inplace=True)
else:
    raise ValueError("Couldn't find the 'Date first added' column. Check Wikipedia table structure.")

# Step 6: Convert 'Date first added' to datetime
df['Date first added'] = pd.to_datetime(df['Date first added'], errors='coerce')

# Step 7: Extract year
df['Year Added'] = df['Date first added'].dt.year

# Step 8: Count additions per year (from 2020 onward)
additions_per_year = df['Year Added'].value_counts().sort_index()
recent_additions = additions_per_year[additions_per_year.index >= 2020]

print("\nAdditions per year since 2020:")
print(recent_additions)

# Step 9: Find the year with the highest number of additions
if not recent_additions.empty:
    max_year = recent_additions.idxmax()
    max_count = recent_additions.max()
    print(f"\nYear with highest additions since 2020: {max_year} ({max_count} companies)")
else:
    print("\nNo additions found since 2020.")

# Additional: Companies in index for more than 20 years
df['Years in Index'] = pd.Timestamp.today().year - df['Year Added']
long_tenure = df[df['Years in Index'] >= 20]

print(f"\nNumber of companies in the index for more than 20 years: {len(long_tenure)}")

# Step 10: Visualization
plt.figure(figsize=(8,5))
recent_additions.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title("S&P 500 Additions per Year (2020 onward)")
plt.xlabel("Year")
plt.ylabel("Number of Additions")
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Question 2: Global Index YTD Returns (final stable version)
import yfinance as yf
import pandas as pd

tickers = {
    "US (S&P 500)": "^GSPC",
    "Hong Kong (Hang Seng)": "^HSI",
    "Australia (ASX 200)": "^AXJO",
    "India (Nifty 50)": "^NSEI",
    "Canada (TSX Composite)": "^GSPTSE",
    "Germany (DAX)": "^GDAXI",
    "UK (FTSE 100)": "^FTSE",
    "Japan (Nikkei 225)": "^N225",
    "Brazil (Ibovespa)": "^BVSP"
}

start_date = "2026-01-01"
end_date = "2026-08-21"

rows = []

for name, ticker in tickers.items():
    try:
        data = yf.download(ticker, start=start_date, end=end_date, progress=False)
        if not data.empty:
            start_price = float(data['Close'].iloc[0])
            end_price = float(data['Close'].iloc[-1])
            ytd_return = (end_price / start_price) - 1
            rows.append({"Index": name, "YTD Return": ytd_return})
        else:
            print(f"⚠️ No data for {name} ({ticker})")
    except Exception as e:
        print(f"❌ Error fetching {name} ({ticker}): {e}")

# Build DataFrame only if rows exist
if rows:
    returns_df = pd.DataFrame(rows)
    returns_df = returns_df.sort_values(by='YTD Return', ascending=False).reset_index(drop=True)

    print("\nYTD Returns (Jan 1 – Aug 21, 2026):")
    print(returns_df)

    # Compare with S&P 500 safely
    sp500_row = returns_df.loc[returns_df['Index'] == "US (S&P 500)"]
    if not sp500_row.empty:
        sp500_return = float(sp500_row['YTD Return'].iloc[0])
        better_than_sp500 = sum(returns_df['YTD Return'] > sp500_return)
        print(f"\nNumber of indexes with better YTD returns than S&P 500: {better_than_sp500}")
    else:
        print("\n⚠️ Could not fetch S&P 500 data, comparison skipped.")
else:
    print("\n⚠️ No returns calculated. All tickers failed.")
```

```python
# Question 4: "Buy the Dip" Strategy Simulation
import yfinance as yf
import pandas as pd

# Step 1: Download S&P 500 data
sp500 = yf.download("^GSPC", start="1950-01-01", progress=False)
close_prices = sp500['Close'].astype(float)
all_time_high = close_prices.cummax()
sp500['Drawdown %'] = (close_prices - all_time_high) / all_time_high * 100
sp500['IsCorrection'] = sp500['Drawdown %'] <= -5

# Step 2: Identify correction periods
correction_periods = []
in_correction = False
start_high = None
low_price = None

for date in sp500.index:
    is_corr = bool(sp500.loc[date, 'IsCorrection'])
    price = float(sp500.loc[date, 'Close'])
    high_price = float(all_time_high.loc[date])

    if is_corr and not in_correction:
        # Start of correction
        in_correction = True
        start_high = high_price
        low_price = price
    elif is_corr and in_correction:
        # Track lowest price during correction
        if price < low_price:
            low_price = price
    elif not is_corr and in_correction:
        # End of correction (recovery)
        end_price = price
        return_pct = (end_price / low_price - 1) * 100
        correction_periods.append({
            "Start High": start_high,
            "Low Price": low_price,
            "Recovery Price": end_price,
            "Return %": return_pct
        })
        in_correction = False

# Step 3: Convert to DataFrame
strategy_df = pd.DataFrame(correction_periods)

print("\n'Buy the Dip' Strategy Returns:")
print(strategy_df.head(10))  # show first 10 corrections

# Step 4: Summary stats
if not strategy_df.empty:
    avg_return = strategy_df['Return %'].mean()
    median_return = strategy_df['Return %'].median()
    print(f"\nAverage Return per Correction: {avg_return:.2f}%")
    print(f"Median Return per Correction: {median_return:.2f}%")
else:
    print("\n⚠️ No corrections found.")
```

```python
# Question 5: Multi-Year Sector Performance Comparison
import yfinance as yf
import pandas as pd

# Step 1: Define sector ETFs
sector_etfs = {
    "Technology (XLK)": "XLK",
    "Healthcare (XLV)": "XLV",
    "Financials (XLF)": "XLF",
    "Energy (XLE)": "XLE",
    "Consumer Discretionary (XLY)": "XLY",
    "Industrials (XLI)": "XLI",
    "Utilities (XLU)": "XLU",
    "Materials (XLB)": "XLB",
    "Real Estate (XLRE)": "XLRE",
    "Communication Services (XLC)": "XLC"
}

# Step 2: Define periods
periods = {
    "YTD (2026)": ("2026-01-01", "2026-08-21"),
    "3-Year": ("2023-01-01", "2026-08-21"),
    "5-Year": ("2021-01-01", "2026-08-21")
}

rows = []

# Step 3: Download data and calculate returns for each sector & period
for name, ticker in sector_etfs.items():
    sector_data = {"Sector": name}
    for label, (start_date, end_date) in periods.items():
        try:
            data = yf.download(ticker, start=start_date, end=end_date, progress=False)
            if not data.empty:
                start_price = data['Close'].iloc[0].squeeze()
                end_price = data['Close'].iloc[-1].squeeze()
                return_pct = (end_price / start_price - 1) * 100
                sector_data[label] = return_pct
            else:
                sector_data[label] = None
        except Exception as e:
            sector_data[label] = None
    rows.append(sector_data)

# Step 4: Build DataFrame
sector_df = pd.DataFrame(rows)

print("\nS&P 500 Sector Performance (YTD, 3-Year, 5-Year):")
print(sector_df)
```

```python
# Question 5: Ranking Summary of Sector Performance
import pandas as pd

# Assuming you already built `sector_df` with YTD, 3-Year, and 5-Year returns

# Step 1: For each period, rank sectors by return
rankings = {}
for period in sector_df.columns:
    if period != "Sector":
        top3 = sector_df.sort_values(by=period, ascending=False).head(3)
        rankings[period] = top3[["Sector", period]]

# Step 2: Print ranking summary
print("\nTop 3 Sectors by Period:")
for period, df in rankings.items():
    print(f"\n{period}:")
    print(df.to_string(index=False))
```
